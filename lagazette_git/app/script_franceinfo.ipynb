```python
import feedparser
from pprint import pprint
import json
```

```python
# National
lien_franceinfo = "https://www.franceinfo.fr/titres.rss"
```

Stocker la rubrique

```python
d = feedparser.parse(lien_franceinfo)
```

```python
with open("../bronze/franceinfo.json", "r") as file:
    file = json.load(file)
```

```python
dict_franceinfo = {}
# le dernier indice de l'ancien fichier converti en int
index = int(list(file.keys())[-1])
for i in d['entries'] : 
# liste d'infos par article
    index += 1
    dict_franceinfo[index] = [i['title'], i['link'], i['summary'], i['published']]
```

```python
file.update(dict_franceinfo)
```
