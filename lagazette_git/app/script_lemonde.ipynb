```python
import feedparser
from pprint import pprint
import json
```

```python
# National
lien_lemonde = "https://www.lemonde.fr/rss/une.xml"
```

Stocker la rubrique

```python
d = feedparser.parse(lien_lemonde)
```

```python
dict_lemonde = {}
index = 0
for i in d['entries'] : 
# liste d'infos par article
    dict_lemonde[index] = [i['title'], i['link'], i['summary'], i['published']]
    index += 1
```
