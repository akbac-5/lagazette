```python
import feedparser
from pprint import pprint
import json
```

```python
# Régional
lien_sudouest = "https://www.sudouest.fr/rss.xml"
```

Stocker la rubrique

```python
d = feedparser.parse(lien_sudouest)
```

```python
dict_sudouest = {}
index = 0
for i in d['entries'] : 
# liste d'infos par article
    dict_sudouest[index] = [i['title'], i['link'], i['summary'], i['published']]
    index += 1
```
